```python
df=pd.read_csv('auto_mpg.csv')
df
```

```python
df.drop('name',axis=1,inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['origin']=le.fit_transform(df['origin'])
df
```

```python
df.isnull().sum()
```

```python
df.fillna(method='ffill',inplace=True)
df.isnull().sum()
```

```python
X=df.iloc[:,1:8]
Y=df[['mpg']]
```

```python
X
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(
X,Y,test_size=0.25,random_state=42)
```

```python
x_train
```

```python
x_test
```

```python
y_train
```

```python
y_test
```

```python
from sklearn.ensemble import RandomForestRegressor
rfr=RandomForestRegressor(n_estimators=100,random_state=0)
rfr.fit(x_train,y_train)
```

```python
y_pred1=rfr.predict(x_test)
y_pred1
```

```python
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
mse_RFR=mean_squared_error(y_test,y_pred1)
mae_RFR=mean_absolute_error(y_test,y_pred1)
R2_RFR=r2_score(y_test,y_pred1)
rmses_RFR=np.sqrt(mse_RFR)
print('mean_squared_error:',mse_RFR)
print('mean_absolute_error:',mae_RFR)
print('root_mean_srured_error:',rmses_RFR)
print('R2_score:',r2_score)
```

```python

mse_SVR=mean_squared_error(y_test,y_pred1)
mae_SVR=mean_absolute_error(y_test,y_pred1)
R2_SVR=np.sqrt(mse_SVR)
rmses_SVR=np.sqrt(mse_SVR)
print('mean_squared_error:',mse_SVR)
print('mean_absolute_error:',mae_SVR)
print('root_mean_srured_error:',rmses_SVR)
print('R2_score:',r2_score)
```

```python

mse_LA=mean_squared_error(y_test,y_pred1)
mae_LA=mean_absolute_error(y_test,y_pred1)
R2_LA=np.sqrt(mse_LA)
rmses_LA=np.sqrt(mse_LA)
print('mean_squared_error:',mse_LA)
print('mean_absolute_error:',mae_LA)
print('root_mean_srured_error:',rmses_LA)
print('R2_score:',r2_score)
```

```python
pd.DataFrame({
    'Regression':[
        'RandomForestRegressor',
        'supportvectorRegression',
        'linearRegression'
    ],
    'mse':[mse_RFR, mse_SVR, mse_LA],
    'mae':[mae_RFR, mae_SVR, mae_LA],
    'rmse':[rmses_RFR,rmses_SVR,rmses_LA]})
```
